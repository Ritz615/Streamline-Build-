# 01 — Dataset Exploration

This notebook explores the ds007169 dataset structure:
- BIDS directory layout
- Subject list
- Channel information
- Event types and label mapping
- Raw EEG visualisation

```python
import sys
sys.path.insert(0, '..')
from src.config import get_config, setup_logging
from src.data_loader import OpenNeuroNBackDataset
import pandas as pd
import matplotlib.pyplot as plt

config = get_config()
setup_logging(config)
print('Config loaded:', config.dataset.id)
```

```python
# Load dataset
ds = OpenNeuroNBackDataset(config=config)
info = ds.get_dataset_info()
print('Dataset info:')
for k, v in info.items():
    print(f'  {k}: {v}')
```

```python
# Subject list
subjects = ds.get_subject_list()
print(f'Found {len(subjects)} subjects: {subjects}')
```

```python
# Load first subject
sub_id = subjects[0]
raw, events_df, label_map = ds.load_subject(sub_id)
print(f'{sub_id}: {raw.info["sfreq"]} Hz, {len(raw.ch_names)} channels, {raw.times[-1]:.1f} s')
print('Label mapping:', label_map)
print('Events sample:')
print(events_df.head(10))
```

```python
# Event class distribution
counts = events_df['label'].value_counts()
print('Label counts:')
print(counts)
counts.plot(kind='bar', color=['#2E86AB','#F6AE2D','#E84855'], title='Event Distribution')
plt.tight_layout()
plt.show()
```

```python
# Plot raw EEG (first 10 seconds)
raw.plot(duration=10, n_channels=8, scalings=dict(eeg=50e-6), title=f'{sub_id} Raw EEG')
plt.show()
```
