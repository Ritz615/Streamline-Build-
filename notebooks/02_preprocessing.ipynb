# 02 — Preprocessing

Demonstrates the EEG preprocessing pipeline step by step.

```python
import sys
sys.path.insert(0, '..')
from src.config import get_config, setup_logging
from src.data_loader import OpenNeuroNBackDataset
from src.preprocessing import preprocess_subject
import matplotlib.pyplot as plt

config = get_config()
setup_logging(config)
```

```python
ds = OpenNeuroNBackDataset(config=config)
subjects = ds.get_subject_list()
sub_id = subjects[0]
raw, events_df, _ = ds.load_subject(sub_id)
print(f'Loaded {sub_id}: {len(raw.ch_names)} channels')
```

```python
# Run preprocessing
raw_clean, report = preprocess_subject(sub_id, raw, config=config, save=False)
print('Preprocessing report:')
for k, v in report.items():
    if k != 'data':
        print(f'  {k}: {v}')
```

```python
# Compare raw vs clean power spectra
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
raw.plot_psd(fmax=50, ax=axes[0], show=False)
axes[0].set_title('Before Preprocessing')
raw_clean.plot_psd(fmax=50, ax=axes[1], show=False)
axes[1].set_title('After Preprocessing')
plt.tight_layout()
plt.show()
```
