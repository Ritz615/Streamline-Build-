# 03 — Feature Analysis

Loads features.csv and performs exploratory analysis.

```python
import sys
sys.path.insert(0, '..')
from src.config import get_config
from src.feature_extraction import load_features, get_feature_columns
from src.feature_selection import select_features
from src.visualization import run_eda, plot_feature_importance
import pandas as pd
import matplotlib.pyplot as plt

config = get_config()
df = load_features(config=config)
print(df.shape)
print(df.head())
```

```python
# Class distribution
print(df['label'].value_counts())
```

```python
# Feature importance
selected, importance_df = select_features(df, config=config)
print('Top 5 features:', selected)
print(importance_df.head(10))
```

```python
# Run full EDA
run_eda(df, config=config)
plot_feature_importance(importance_df, config=config)
print('EDA figures saved to results/figures/')
```

```python
# Show theta/beta ratio by class
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(8,5))
for cls, color in [('LOW','#2E86AB'),('MODERATE','#F6AE2D'),('HIGH','#E84855')]:
    vals = df[df['label']==cls]['theta_beta_ratio'].dropna()
    ax.hist(vals, bins=30, alpha=0.5, label=cls, color=color, density=True)
ax.set_xlabel('Theta/Beta Ratio')
ax.set_title('Theta/Beta Ratio Distribution by Class')
ax.legend()
plt.show()
```
