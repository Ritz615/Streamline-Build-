# 04 — Fuzzy Model

Demonstrates the fuzzy inference system interactively.

```python
import sys
sys.path.insert(0, '..')
from src.config import get_config
from src.feature_extraction import load_features, get_feature_columns
from src.feature_selection import get_top_features
from src.fuzzy_classifier import FuzzyClassifier
from src.explainability import explain_fuzzy
from sklearn.preprocessing import LabelEncoder
import numpy as np

config = get_config()
df = load_features(config=config)
feat_cols = get_feature_columns(df)
X = df[feat_cols].values
le = LabelEncoder()
y = le.fit_transform(df['label'].values)
```

```python
# Get top features for fuzzy
top_feats = get_top_features(n=5, config=config)
feat_idx = [feat_cols.index(f) for f in top_feats if f in feat_cols]
X_fuzzy = X[:, feat_idx]
actual_feats = [feat_cols[i] for i in feat_idx]
print('Fuzzy features:', actual_feats)
```

```python
# Train fuzzy
clf = FuzzyClassifier(config=config)
clf.train(X_fuzzy, y, actual_feats)
print(f'Trained with {len(clf.rules)} rules')
```

```python
# Make a prediction
test_x = X_fuzzy[0]
result = clf.predict_single(test_x)
print('Prediction:', result['predicted_class'])
print('Fuzzy score:', result['fuzzy_score'])
print('Activated rules:', len(result['activated_rules']))
```

```python
# Print full explanation
explanation = explain_fuzzy(result, test_x, actual_feats)
print(explanation)
```

```python
# Print all rules
print('\nAll fuzzy rules:')
for rule in clf.rules:
    print(f"  Rule {rule['rule_number']}: {rule['rule_text']}")
```
